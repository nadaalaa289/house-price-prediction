```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parent
RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"

translation_df = pd.read_csv(
    RAW_DATA_DIR / "product_category_name_translation.csv"
)

products_df = pd.read_csv(
    RAW_DATA_DIR / "olist_products_dataset.csv"
)
```

```python
print("Translation Shape:", translation_df.shape)
print("Products Shape:", products_df.shape)
```

```python
print("Translation Columns:")
print(translation_df.columns)

print("\nProducts Columns:")
print(products_df.columns)
```

```python
display(translation_df.head())
display(products_df.head())
```

```python
# Check duplicated category names
duplicate_categories = translation_df[
    translation_df["product_category_name"].duplicated(keep=False)
]

print("Number of duplicated categories:", len(duplicate_categories))

display(duplicate_categories)
```

```python
duplicate_translations = translation_df[
    translation_df["product_category_name_english"].duplicated(keep=False)
]

print("Number of duplicated translations:", len(duplicate_translations))

display(duplicate_translations)
```

```python
print("Category duplicates: 0")
print("Translation duplicates: 0")
print("Result: No duplicate values found.")
```

```python
product_categories = set(
    products_df["product_category_name"].dropna()
)

translated_categories = set(
    translation_df["product_category_name"].dropna()
)
```

```python
missing_translations = product_categories - translated_categories

print("Number of categories without translation:", len(missing_translations))
print("Categories without translation:")

for category in sorted(missing_translations):
    print(category)
```

```python
for category in missing_translations:
    count = products_df[
        products_df["product_category_name"] == category
    ].shape[0]

    print(f"{category}: {count} products")
```

Missing Category Translations

Two categories were found in the products dataset but not in the translation dataset:

1. pc_gamer: 3 products
2. portateis_cozinha_e_preparadores_de_alimentos: 10 products

Total affected products: 13.

These categories need further investigation.

```python
print("Missing Values in Translation Dataset:")
print(translation_df.isnull().sum())

print("\nMissing Values in Products Dataset:")
print(products_df.isnull().sum())
```

```python
missing_rows = products_df[
    products_df["product_category_name"].isnull()
]

print("Number of rows with missing category:", len(missing_rows))

display(missing_rows.head())
```
